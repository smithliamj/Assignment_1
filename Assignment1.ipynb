**2. Are there any null values in the Grc_df? If yes, then in which columns and how many?
Finally, handle these null values using any strategy shown during the labs.**

```python
# Find number of null values
Grc_df.isna().sum()
```

Column Item_Weight has 818 null values and column Outlet_Size has 1439 null values

```python
# Replace the null values in Item_Weight with the median
Grc_df["Item_Weight"] = Grc_df["Item_Weight"].fillna(
    Grc_df["Item_Weight"].median()
)
# Replace the null value in Outlet_Size with the most common
Grc_df["Outlet_Size"] = Grc_df["Outlet_Size"].fillna(
    Grc_df["Outlet_Size"].mode()[0]
)
```

**4. How many unique Item Fat Content types are in the Grc_df? List them. Do you see any
issues with the Item Fat Content types? If yes, then handle this issue.**

```python
# Find number of unique Fat Content types
Grc_df["Item_Fat_Content"].unique()
```

There are five Item Fat Content types: 'Low Fat', 'Regular', 'low fat', 'LF' and 'reg'.  
Issue is that there should only be two Fat Content types, but the Low Fat type is recorded in three different ways (Low Fat, low fat and LF) and the Regular type is recorded in two different ways (Regular, reg)

```python
# Replace duplicate values
Grc_df["Item_Fat_Content"] = Grc_df["Item_Fat_Content"].replace('low fat', 'Low Fat')
Grc_df["Item_Fat_Content"] = Grc_df["Item_Fat_Content"].replace('LF', 'Low Fat')
Grc_df["Item_Fat_Content"] = Grc_df["Item_Fat_Content"].replace('reg', 'Regular')
# Check only two Fat Context types after change
Grc_df.Item_Fat_Content.unique()
```

**7. Using Seaborn (“ggplot style”) create a (2,1) subplot of a box plot showing 5-point summary of the column Item_MRP for SupType_1 and SupType_2. Which Outlet Type has a higher median MRP? Are there any outliers?**

```python
# Use ggplot style
plt.style.use("ggplot")
# Create 2 subplots arranged vertically
fig, axes = plt.subplots(2, 1, figsize=(10, 8))
# Create box plot for Supermarket Type1
sns.boxplot(x=SupType_1["Item_MRP"], ax=axes[0])
axes[0].set_title("Supermarket Type1")
# Create box plot for Supermarket Type2
sns.boxplot(x=SupType_2["Item_MRP"], ax=axes[1])
axes[1].set_title("Supermarket Type2")
# Adjust spacing between the plots
plt.tight_layout()
# Display the plots
plt.show()
```

**10. Cut the column Item_Weight of Grc_Concat_df into 10 buckets and compute each bucket's
mean, minimum, maximum, and count.**

```python
# Cut Item_Weight into 10 buckets
Grc_Concat_df["Weight_Bucket"] = pd.cut(Grc_Concat_df["Item_Weight"], bins=10)
# Group the rows by weight interval and calculate the mean, minimum, maximum, and count
weight_stats = Grc_Concat_df.groupby(
    "Weight_Bucket",
    observed=False
)["Item_Weight"].agg(
    ["mean", "min", "max", "count"]
)

# Display the statistics for each of the 10 intervals
weight_stats
```
